```python
import pandas as pd
import json
import os
```

```python
with open(r'C:\Users\bhara\OneDrive\Desktop\Guvi\Uber Eats\orders.json', encoding='utf-8') as f:
    data = json.load(f)
```

```python
orders_df = pd.json_normalize(data)

print("Orders Shape:", orders_df.shape)
orders_df.head()
```

```python
orders_df['order_date'] = pd.to_datetime(orders_df['order_date'], errors='coerce')
orders_df['order_value'] = pd.to_numeric(orders_df['order_value'], errors='coerce')
```

```python
os.makedirs('data/cleaned', exist_ok=True)
orders_df.to_csv('data/cleaned/orders_cleaned.csv', index=False)
print("Orders saved!")
```
